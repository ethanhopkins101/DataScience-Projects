```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
path= 'creditcard.csv'
data= pd.read_csv(path)
df= data.copy()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split as tts
inputs= df.drop('Class', axis=1)
target= df[['Class']]
x_train, x_test, y_train, y_test= tts(inputs, target, stratify= target, random_state= 22, test_size= .2)
std= StandardScaler()
x_train[['Time','Amount']]= std.fit_transform(x_train[['Time','Amount']])
x_test[['Time','Amount']]= std.transform(x_test[['Time','Amount']])
```

```python
from sklearn.linear_model import LogisticRegression
logit= LogisticRegression(class_weight= 'balanced')
logit.fit(x_train, y_train)
```

```python
from sklearn.metrics import classification_report as cr
print(cr(y_test, logit.predict(x_test)))
```

```python
import optuna
import xgboost as xgb
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
from sklearn.model_selection import RepeatedStratifiedKFold, cross_val_score

def objective(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'scale_pos_weight': trial.suggest_float('scale_pos_weight', 1, 100), 
        'use_label_encoder': False,
        'eval_metric': 'logloss'
    }
    
    model = xgb.XGBClassifier(**params, random_state=42)
    cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=42)
    
    # We use f1 score to balance precision and recall during optimization
    score = cross_val_score(model, x_train, y_train, cv=cv, scoring='f1', n_jobs=-1).mean()
    return score

# 1. Initialize and run the study
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=20) # Start with 20 trials to save time

# 2. Train the final model with best parameters
best_model = xgb.XGBClassifier(**study.best_params, random_state=42)
best_model.fit(x_train, y_train)

# 3. Evaluate
y_pred = best_model.predict(x_test)
print("Best Params:", study.best_params)
print(classification_report(y_test, y_pred))

# 4. Plot Confusion Matrix
ConfusionMatrixDisplay.from_estimator(best_model, x_test, y_test, cmap='BuPu')
plt.show()
```

```python
# Feed the best params to a new model
best_model = xgb.XGBClassifier(**study.best_params, random_state=42)
best_model.fit(x_train, y_train)

# Test and Report
y_pred = best_model.predict(x_test)
print(classification_report(y_test, y_pred))
```

```python
# Plot Confusion Matrix with 'BuPu' (Blue-Purple)
sns.reset_defaults()
fig, ax = plt.subplots(figsize=(8, 6))
ConfusionMatrixDisplay.from_estimator(
    best_model, x_test, y_test, cmap='BuPu', ax=ax
)
plt.title("Confusion Matrix - Optimized XGBoost")
plt.show()
```
